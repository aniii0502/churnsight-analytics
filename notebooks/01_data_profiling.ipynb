```python
import pandas as pd

# 1. Load the raw dataset
# Assuming the notebook is in notebooks/ and the CSV is in data/
df = pd.read_csv('../data/WA_Fn-UseC_-Telco-Customer-Churn.csv')
```

```python

# 2. Check dataset shape (rows vs. columns)
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")
```

```python

# 3. Inspect data types to flag parsing needs (e.g., dates loaded as objects)
print("--- Data Types ---")
print(df.dtypes, "\n")
```

```python

# 4. Count missing values per column to identify right-censoring vs. bad data
print("--- Missing Values ---")
print(df.isnull().sum(), "\n")
```

```python
print(df.columns)
```

```python
# Replace 'customerID' and other keys on the left with the actual column names from your CSV
df = df.rename(columns={
    'customerID': 'user_id', 
    'tenure': 'tenure_months', # Example mappings depending on your raw data
    'MonthlyCharges': 'monthly_spend'
})

# Verify the rename worked
print(df.columns)
```

```python

# 5. Check for duplicate user_ids which would break cohort and RFM grouping
duplicate_users = df.duplicated(subset=['user_id']).sum()
print(f"--- Duplicate user_ids ---")
print(f"Found {duplicate_users} duplicate records.")
```

```python

# Optional: Preview the exact duplicate rows if any exist
if duplicate_users > 0:
    display(df[df.duplicated(subset=['user_id'], keep=False)].sort_values('user_id'))
```
